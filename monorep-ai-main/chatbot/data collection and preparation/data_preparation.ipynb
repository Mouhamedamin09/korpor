```python
import json
import pandas as pd
```

### Combine the data into three files

```python
def concat_json_files(file1, file2, file3, output_file):
    data = []
    
    for file in [file1, file2, file3]:
        with open(file, 'r', encoding='utf-8') as f:
            data.extend(json.load(f))  # Assuming JSON files contain lists
    
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(data, f, indent=4, ensure_ascii=False)

# Example usage
concat_json_files('real_state_en.json', 'registration_en.json', 'laws_en.json', 'data_en.json')
concat_json_files('real_state_fr.json', 'registration_fr.json', 'laws_fr.json', 'data_fr.json')
concat_json_files('real_state_ar.json', 'registration_ar.json', 'laws_ar.json', 'data_ar.json')
```

## Changing data format 

```python
def json_to_dataframe(file):
    with open(file, 'r', encoding='utf-8') as file:
        data = json.load(file)
    # Convert to DataFrame
    df = pd.DataFrame(data)
    return df
```

```python
df_en=json_to_dataframe('data_en.json')
df_en.head()
```

```python
df_fr=json_to_dataframe('data_fr.json')
df_fr.head()
```

```python
df_ar=json_to_dataframe('data_ar.json')
df_ar.head()
```

```python
def format_answer(group):
    combined = []
    for index, row in group.iterrows():
        combined.append(f"{row['subtopic']}:\n- {row['answer']}")
    return '\n\n'.join(combined)
```

```python
columns_order=['section','question','answer']
```

```python
data_en_grouped = df_en.groupby(['question', 'section']).apply(format_answer).reset_index(name='answer')
data_en_grouped = data_en_grouped[columns_order]
data_en_grouped.head()
```

```python
data_ar_grouped = df_ar.groupby(['question', 'section']).apply(format_answer).reset_index(name='answer')
data_ar_grouped=data_ar_grouped[columns_order]
data_ar_grouped.head()
```

```python
data_fr_grouped = df_fr.groupby(['question', 'section']).apply(format_answer).reset_index(name='answer')
data_fr_grouped=data_fr_grouped[columns_order]
data_fr_grouped.head()
```

```python
# Convert DataFrames to JSON strings with special characters preserved
data_en_json = data_en_grouped.to_dict(orient='records')
data_fr_json = data_fr_grouped.to_dict(orient='records')
data_ar_json = data_ar_grouped.to_dict(orient='records')
with open('data_en.json', 'w', encoding='utf-8') as f:
    json.dump(data_en_json, f, ensure_ascii=False, indent=4)

with open('data_fr.json', 'w', encoding='utf-8') as f:
    json.dump(data_fr_json, f, ensure_ascii=False, indent=4)

with open('data_ar.json', 'w', encoding='utf-8') as f:
    json.dump(data_ar_json,f,ensure_ascii=False,indent=4)
```
