```python
import pandas as pd
```

```python
df1=pd.read_csv('municipality_governorate_description_22.csv')
df2=pd.read_csv('municipality_governorate_description_24.csv')
print("Number of unique municipalities in df1: ", df1['municipality'].nunique())
print("Number of unique governorates in df1: ",df1['governorate'].nunique())
print("Number of unique municipalities in df2: ",df2['municipality'].nunique())
print("Number of unique governorates in df2: ",df2['governorate'].nunique())
```

```python
# Identify municipalities in df1 that are not in df2
missing_municipalities = df1[~df1['municipality'].isin(df2['municipality'])]
# Append the missing municipalities to df2
df = pd.concat([df2, missing_municipalities], ignore_index=True) #ingore index resets the indexing
print("Updated number of unique municipalities: ", df['municipality'].nunique())
print("Updated number of unique governorates: ", df['governorate'].nunique())
```
